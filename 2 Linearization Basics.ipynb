#Linearization Basics

In this notebook, we will go through another example of converting an optimization problem with an absolute value into a linear optimization model. The model we will look at is as follows:

*Minimize*:  
$ 5X + 2|Y| $

*Subject* to:  
$ X + Y \geq 9 $

This time, the absolute value is in the objective function. As written, this would be a challenging nonlinear model. Let's convert this to a linear optimization model using the approach we saw in class.

First, let's get a sense for the behavior of the nonlinear model. Using the sliders below to set the decision variable values, what is the smallest you can make the objective function below, while not violating the constraint? (Don't worry about the animation code, unless you are interested!)




```python
#@title Nonlinear Model

# Import required libraries
import ipywidgets as widgets
from IPython.display import display, Math

# Create sliders for decision variables
x_slider = widgets.FloatSlider(value=0.0, min=-20.0, max=20.0, step=0.1, description="X")
y_slider = widgets.FloatSlider(value=0.0, min=-20.0, max=20.0, step=0.1, description="Y")

# Output widget for displaying equations
output = widgets.Output()

# Function to dynamically update the equations
def update_equations(x, y):
    with output:
        # Clear previous output
        output.clear_output(wait=True)

        # Compute the absolute value of Y
        abs_y = abs(y)

        # Compute the objective function value
        Z = 5 * x + 2 * abs_y

        # Compute the constraint value
        constraint_value = x + y

        # Determine the constraint status
        constraint_status = r"\textcolor{green}{Satisfied}" if constraint_value >= 9 else r"\textcolor{red}{Violated}"

        # Create LaTeX strings for the objective function and constraint
        objective_latex = rf"Minimize \quad 5X + 2|Y| = 5({x:.1f}) + 2|{y:.1f}| = {Z:.1f}"
        constraint_latex = rf"Subject \, to: \, X + Y \geq 9" \
                           rf" \quad \text{{where }} {x:.1f} + {y:.1f} = {constraint_value:.1f}" \
                           rf" \quad \textbf{{{constraint_status}}}"

        # Display the equations
        display(Math(objective_latex))
        display(Math(constraint_latex))

# Create an interactive widget
interactive_display = widgets.interactive(update_equations, x=x_slider, y=y_slider)

# Arrange the output and sliders in a vertical layout
layout = widgets.VBox([output, x_slider, y_slider])

# Display the layout
display(layout)

# Initialize the equations
update_equations(0.0, 0.0)
```

First, we will make a variable change: $|Y|=MAX(Y, -Y)=V$.  Then we will enforce this with two new constraints: $V\geq Y$ and $V\geq -Y$ - in other words, $V \geq MAX(Y, -Y)$. This gives us a new model:

*Minimize*:  
$ 5X + 2V $

*Subject* to:  
$V \geq Y$

$V \geq -Y$

$ X + Y \geq 9 $

Let's double check and make sure these new constraints work. At optimality, would it would be impossible for $V$ to be strictly greater than both $Y$ and $-Y$ - in other words, would it be possible for $V$ to not equal $MAX(Y, -Y)$? No - if it was, we could simply reduce V and improve the objective function (since we are minimizing) until our value of $V$ ran up against either $Y$ or $-Y$, in which case $MAX(Y, -Y)=V$.

Now try adjusting the sliders below to test this new model: how small can you make the objective function while satisfying the constraints? How do the results compare to the original model?

```python
#@title Linearized Model

# Import required libraries
import ipywidgets as widgets
from IPython.display import display, Math

# Create sliders for decision variables
x_slider = widgets.FloatSlider(value=0.0, min=-20.0, max=20.0, step=0.1, description="X")
y_slider = widgets.FloatSlider(value=0.0, min=-20.0, max=20.0, step=0.1, description="Y")
v_slider = widgets.FloatSlider(value=0.0, min=0.0, max=20.0, step=0.1, description="V")

# Output widget for displaying equations
output = widgets.Output()

# Function to dynamically update the equations
def update_equations(x, y, v):
    with output:
        # Clear previous output
        output.clear_output(wait=True)

        # Compute the objective function value
        Z = 5 * x + 2 * v

        # Compute constraint values
        constraint_1_status = r"\textcolor{green}{Satisfied}" if v >= y else r"\textcolor{red}{Violated}"
        constraint_2_status = r"\textcolor{green}{Satisfied}" if v >= -y else r"\textcolor{red}{Violated}"
        constraint_3_value = x + y
        constraint_3_status = r"\textcolor{green}{Satisfied}" if constraint_3_value >= 9 else r"\textcolor{red}{Violated}"

        # Create LaTeX strings for the objective function and constraints
        objective_latex = rf"Minimize \, Z = 5X + 2V = 5({x:.1f}) + 2({v:.1f}) = {Z:.1f}"
        constraint_1_latex = rf"V \geq Y \quad \text{{where }} {v:.1f} \geq {y:.1f} \quad \textbf{{{constraint_1_status}}}"
        constraint_2_latex = rf"V \geq -Y \quad \text{{where }} {v:.1f} \geq -({y:.1f}) = {-y:.1f} \quad \textbf{{{constraint_2_status}}}"
        constraint_3_latex = rf"X + Y \geq 9 \quad \text{{where }} {x:.1f} + {y:.1f} = {constraint_3_value:.1f} \quad \textbf{{{constraint_3_status}}}"

        # Display the equations
        display(Math(objective_latex))
        display(Math(constraint_1_latex))
        display(Math(constraint_2_latex))
        display(Math(constraint_3_latex))

# Create an interactive widget
interactive_display = widgets.interactive(update_equations, x=x_slider, y=y_slider, v=v_slider)

# Arrange the output and sliders in a vertical layout
layout = widgets.VBox([output, x_slider, y_slider, v_slider])

# Display the layout
display(layout)

# Initialize the equations
update_equations(0.0, 0.0, 0.0)
```
