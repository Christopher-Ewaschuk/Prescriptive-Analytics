# Formulating a Linear Optimization Model

In the last lesson, we looked at the process of formulating a linear optimization model: we start with a set of decision variables that we can adjust, decide on an objective function depending on those variables that we are trying to minimize, and decide on the contraints restricting the values and relationships of those variables.

We then put together a model of transit ridership: we want to maximize the total ridership (the objective function) by offering a certain number of trips for heavy rail, light rail, bus rapid transit (BRT) and bus (our decision variables $x_1, x_2, x_3, x_4$ respectively), subject to a constraint that the total cost of all trips falls within our budget. This gave us the following model:



**Maximize Total Ridership:**

$
400x_1 + 125x_2 + 60x_3 + 40x_4
$

**Subject to: costs being less than or equal to budget**

$
 20x_1 + 8x_2 + 4x_3 + 3x_4 \leq 500
$

**and:  number of trips being non-negative:**

$
x_1, x_2, x_3, x_4 \geq 0
$

The interactive formula below allows you to adjust each of the decision variables and see the objective value (on the first line), and the cost (on the second line). When you go over budget, you will get a red "Violated" message. By adjusting the sliders, see how high you can get the total ridership without violating the constraint. Note that you cannot go below zero trips for any mode of transport. While there is no mathematical limit to the number of trips for any given mode satisfied, you can only go so high on each slider - however, you will see that you can add enough trips on any given slider that the constraint can eventually be violated with only trips from that single mode. You don't need to worry about the details of the animation code, unless you are interested!



```python
#@title Ridership Model

# Import necessary libraries
import ipywidgets as widgets
from IPython.display import display, Math

# Coefficients for the linear program
objective_coeffs = [400, 125, 60, 40]  # Coefficients of the objective function
constraint_coeffs = [20, 8, 4, 3]      # Coefficients of the constraint
constraint_bound = 500                 # Constraint bound

# Create float sliders for the decision variables x1, x2, x3, and x4
x1_slider = widgets.FloatSlider(value=0.0, min=0.0, max=30.0, step=0.1, description="Heavy Rail x₁")
x2_slider = widgets.FloatSlider(value=0.0, min=0.0, max=70.0, step=0.1, description="Light Rail x₂")
x3_slider = widgets.FloatSlider(value=0.0, min=0.0, max=130.0, step=0.1, description="BRT        x₃")
x4_slider = widgets.FloatSlider(value=0.0, min=0.0, max=170.0, step=0.1, description="Bus        x₄")

# Output widget for displaying equations
output = widgets.Output()

# Function to dynamically update and display equations
def update_equations(x1, x2, x3, x4):
    with output:
        # Clear previous output
        output.clear_output(wait=True)

        # Compute objective function value
        Z = (objective_coeffs[0] * x1 +
             objective_coeffs[1] * x2 +
             objective_coeffs[2] * x3 +
             objective_coeffs[3] * x4)

        # Compute constraint function value
        constraint_value = (constraint_coeffs[0] * x1 +
                            constraint_coeffs[1] * x2 +
                            constraint_coeffs[2] * x3 +
                            constraint_coeffs[3] * x4)

        # Determine constraint status
        if constraint_value <= constraint_bound:
            constraint_status = r"\textcolor{green}{Satisfied}"
        else:
            constraint_status = r"\textcolor{red}{Violated}"

        # Create LaTeX strings
        objective_latex = rf"Maximize \,Ridership \quad 400x_1 + 125x_2 + 60x_3 + 40x_4 = " \
                          rf"400({x1:.1f}) + 125({x2:.1f}) + 60({x3:.1f}) + 40({x4:.1f}) = {Z:.1f}"

        constraint_latex = rf"\quad \quad s. t.  \quad  20x_1 + 8x_2 + 4x_3 + 3x_4 \leq {constraint_bound}" \
                           rf" \quad \text{{where }} 20({x1:.1f}) + 8({x2:.1f}) + 4({x3:.1f}) + 3({x4:.1f}) = {constraint_value:.1f}" \
                           rf" \quad \textbf{{{constraint_status}}}"

        # Display equations
        display(Math(objective_latex))
        display(Math(constraint_latex))

# Create an interactive widget
interactive_display = widgets.interactive(update_equations,
                                          x1=x1_slider,
                                          x2=x2_slider,
                                          x3=x3_slider,
                                          x4=x4_slider)

# Arrange equations above sliders
layout = widgets.VBox([output, x1_slider, x2_slider, x3_slider, x4_slider])

# Display the layout
display(layout)

# Initialize the equations
update_equations(0.0, 0.0, 0.0, 0.0)
```

We can adjust our model to include further constraints. For example, suppose that we want a minimum of 10 bus trips offered, and a minimum of 300 BRT riders.We can implement those constraints as discussed in the lesson as $x_2 \geq 10$ and $60x_3 \geq 300$ respectively. We can now add these to the full model.

**Maximize Total Ridership:**

$
400x_1 + 125x_2 + 60x_3 + 40x_4
$

**Subject to: costs being less than or equal to budget**

$
 20x_1 + 8x_2 + 4x_3 + 3x_4 \leq 500
$


**and: at least 10 bus trips:**

$
x_2 \geq 10
$


**and:  at least 300 bus riders:**

$
60x_3 \geq 300
$



**and:  number of trips being non-negative:**

$
x_1, x_2, x_3, x_4 \geq 0
$

The interactive set of equations below shows this model with its additional constraints. Try adjusting the sliders again to see how high you can get the objective function with out violating the constraints. How has your solution changed from the original model?

```python
#@title Transit Model With Additional Constraints
# Import necessary libraries
import ipywidgets as widgets
from IPython.display import display, Math

# Coefficients for the linear program
objective_coeffs = [400, 125, 60, 40]  # Coefficients of the objective function
constraint_coeffs = [20, 8, 4, 3]      # Coefficients of the constraint
constraint_bound = 500                 # Constraint bound
constraint_lower_bound_x2 = 10         # Bound for x2 constraint
constraint_lower_bound_4x3 = 300        # Bound for 4x3 constraint

# Create float sliders for the decision variables x1, x2, x3, and x4
x1_slider = widgets.FloatSlider(value=0.0, min=0.0, max=30.0, step=0.1, description="Heavy Rail x₁")
x2_slider = widgets.FloatSlider(value=0.0, min=0.0, max=70.0, step=0.1, description="Light Rail x₂")
x3_slider = widgets.FloatSlider(value=0.0, min=0.0, max=130.0, step=0.1, description="BRT        x₃")
x4_slider = widgets.FloatSlider(value=0.0, min=0.0, max=170.0, step=0.1, description="Bus        x₄")

# Output widget for displaying equations
output = widgets.Output()

# Function to dynamically update and display equations
def update_equations(x1, x2, x3, x4):
    with output:
        # Clear previous output
        output.clear_output(wait=True)

        # Compute objective function value
        Z = (objective_coeffs[0] * x1 +
             objective_coeffs[1] * x2 +
             objective_coeffs[2] * x3 +
             objective_coeffs[3] * x4)

        # Compute main constraint function value
        constraint_value = (constraint_coeffs[0] * x1 +
                            constraint_coeffs[1] * x2 +
                            constraint_coeffs[2] * x3 +
                            constraint_coeffs[3] * x4)

        # Compute additional constraints
        constraint_status_x2 = r"\textcolor{green}{Satisfied}" if x2 >= constraint_lower_bound_x2 else r"\textcolor{red}{Violated}"
        constraint_status_4x3 = r"\textcolor{green}{Satisfied}" if 60 * x3 >= constraint_lower_bound_4x3 else r"\textcolor{red}{Violated}"

        # Determine main constraint status
        constraint_status_main = r"\textcolor{green}{Satisfied}" if constraint_value <= constraint_bound else r"\textcolor{red}{Violated}"

        # Create LaTeX strings
        objective_latex = rf"Maximize \,Ridership \quad 400x_1 + 125x_2 + 60x_3 + 40x_4 = " \
                          rf"400({x1:.1f}) + 125({x2:.1f}) + 60({x3:.1f}) + 40({x4:.1f}) = {Z:.1f}"

        constraint_latex_main = rf"\quad \quad s. t.  \quad  20x_1 + 8x_2 + 4x_3 + 3x_4 \leq {constraint_bound}" \
                                 rf" \quad \text{{where }} 20({x1:.1f}) + 8({x2:.1f}) + 4({x3:.1f}) + 3({x4:.1f}) = {constraint_value:.1f}" \
                                 rf" \quad \textbf{{{constraint_status_main}}}"

        constraint_latex_x2 = rf"\,\quad\quad\quad\quad x_2 \geq {constraint_lower_bound_x2}" \
                              rf" \quad \text{{where }} x_2 = {x2:.1f}" \
                              rf" \quad \textbf{{{constraint_status_x2}}}"

        constraint_latex_4x3 = rf"\,\quad\quad\quad\quad60x_3 \geq {constraint_lower_bound_4x3}" \
                               rf" \quad \text{{where }} 60({x3:.1f}) = {60 * x3:.1f}" \
                               rf" \quad \textbf{{{constraint_status_4x3}}}"

        # Display equations
        display(Math(objective_latex))
        display(Math(constraint_latex_main))
        display(Math(constraint_latex_x2))
        display(Math(constraint_latex_4x3))

# Create an interactive widget
interactive_display = widgets.interactive(update_equations,
                                          x1=x1_slider,
                                          x2=x2_slider,
                                          x3=x3_slider,
                                          x4=x4_slider)

# Arrange equations above sliders
layout = widgets.VBox([output, x1_slider, x2_slider, x3_slider, x4_slider])

# Display the layout
display(layout)

# Initialize the equations
update_equations(0.0, 0.0, 0.0, 0.0)
```
